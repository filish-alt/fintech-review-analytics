# Sentiment and Thematic Analysis

This notebook covers the exploratory data analysis for Task 2, focusing on quantifying sentiment and extracting overarching themes from Ethiopian banking app reviews.

## 1. Tool Selection Rationale

**Sentiment Model:** `VADER` (Valence Aware Dictionary and sEntiment Reasoner) was selected over `distilbert-base-uncased-finetuned-sst-2-english`.
- **Rationale:** VADER is specifically optimized for social media and short texts (like app reviews). It handles emojis, slang, and punctuation effectively without the need for large neural network dependencies or GPUs. It is extremely fast and provides highly interpretable compound scores.

**NLP Preprocessing & Theme Extraction:** We use `NLTK` for tokenization and lemmatization, and `scikit-learn`'s `TfidfVectorizer` for extracting prominent keywords.

## 2. Thematic Grouping Logic

Based on the initial keyword extraction (TF-IDF), we group terms into the following **5 Overarching Themes**:

1. **App Performance**: Captures stability and speed issues.
   - *Keywords*: slow, crash, bug, update, loading, open, working, error, network.
2. **UI & Design**: Captures user experience with the interface.
   - *Keywords*: interface, design, look, easy, simple, user friendly, navigation.
3. **Customer Support**: Captures human interactions and help.
   - *Keywords*: service, support, branch, call, response, rude, help, agent.
4. **Transactions & Transfers**: Captures the core banking functions.
   - *Keywords*: transfer, money, send, receive, transaction, balance, fee, charge, payment.
5. **Account Access**: Captures authentication and onboarding.
   - *Keywords*: login, password, register, otp, code, access, verification, pin.

Any review that does not strongly match these keywords is categorized as **General/Other**.
